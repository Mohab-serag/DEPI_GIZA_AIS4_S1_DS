```python
###############################-Frist problem: Are You Playing Banjo?######################################
def are_you_playing_banjo(name):
    if name[0] == 'R' or name[0] == 'r':
        return name + " plays banjo"
    else:
        return name + " does not play banjo"
print(are_you_playing_banjo("Rami"))
print(are_you_playing_banjo("mohab"))
print(are_you_playing_banjo("rambo"))


```

```python
#############################-Second problem: Reversing Words in a String##########################
def reverse(st):
    words = st.split()
    words.reverse()
    return " ".join(words)
print(reverse("Hello World"))
print(reverse("Hi There   "))

```
