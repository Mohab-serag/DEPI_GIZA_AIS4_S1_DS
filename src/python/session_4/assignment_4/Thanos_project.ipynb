```python
def thanos_project(folder_name="thanos_folder", files_count=10):
    import os, random

    if not os.path.exists(folder_name):
        os.mkdir(folder_name)

    for i in range(files_count):
        with open(os.path.join(folder_name, f"file_{i}.txt"), "w") as f:
            f.write("I am inevitable")

    files = [
        f for f in os.listdir(folder_name)
        if os.path.isfile(os.path.join(folder_name, f))
    ]

    print("Files before delete:", len(files))

    for file in random.sample(files, len(files)//2):
        os.remove(os.path.join(folder_name, file))

    remaining = os.listdir(folder_name)
    print("Files after delete:", len(remaining))
```

```python
thanos_project()
```
