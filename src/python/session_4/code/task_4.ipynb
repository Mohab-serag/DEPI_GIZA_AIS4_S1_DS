```python
import os
import random
def thanos_project(folder_name, num_files):
    # 1- open / create folder
    if not os.path.exists(folder_name):
        os.mkdir(folder_name)
    # 2- create number of files inside it
    for i in range(num_files):
        file_path = os.path.join(folder_name, f"file_{i}.txt")
        with open(file_path, "w") as f:
            f.write("I am inevitable")
    # 3- check the len of files in that folder
    files = os.listdir(folder_name)
    print("Number of files before delete:", len(files))
    # 4- delete random half of files
    half = len(files) // 2
    files_to_delete = random.sample(files, half)
    for file in files_to_delete:
        os.remove(os.path.join(folder_name, file))
    # 5- check the result
    final_files = os.listdir(folder_name)
    print("Number of files after delete:", len(final_files))
# Example call
thanos_project("thanos_folder", 10)
 
```
